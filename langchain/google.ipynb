```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-1.5-flash",
    temperature=0,
    max_tokens=1024,
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
messages = [
    (
        "system",
        "How are you",
    ),
    ("human", "I love programming."),
]
ai_msg = llm.invoke(messages)
```

```python
print(ai_msg.content)
```
